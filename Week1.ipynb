```python
# Tic-Tac-Toe Game

# Human = X
# Computer = O

board = [' ' for _ in range(9)]


def display_board():

    print()
    print(" ", board[0], "|", board[1], "|", board[2])
    print("---+---+---")
    print(" ", board[3], "|", board[4], "|", board[5])
    print("---+---+---")
    print(" ", board[6], "|", board[7], "|", board[8])
    print()


def check_winner(player):

    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:

        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


def board_full():

    return ' ' not in board


def computer_move():

    for i in range(9):

        if board[i] == ' ':
            board[i] = 'O'
            return


print("TIC-TAC-TOE")
print("Human = X")
print("Computer = O")


while True:

    display_board()

    # Human Move
    while True:

        position = int(input("Enter your position (1-9): "))

        if position < 1 or position > 9:
            print("Enter a number between 1 and 9.")

        elif board[position - 1] != ' ':
            print("Position already occupied.")

        else:
            board[position - 1] = 'X'
            break

    # Check Human Win
    if check_winner('X'):
        display_board()
        print("Human wins!")
        break

    # Check Draw
    if board_full():
        display_board()
        print("Game Draw!")
        break

    # Computer Move
    computer_move()

    print("Computer made its move.")

    # Check Computer Win
    if check_winner('O'):
        display_board()
        print("Computer wins!")
        break

    # Check Draw
    if board_full():
        display_board()
        print("Game Draw!")
        break
```

```python
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)


def dfs(state, limit, path):
    if state == goal:
        return path

    if limit == 0:
        return None

    zero = state.index(0)
    row, col = zero // 3, zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            new_state = list(state)

            new_state[zero], new_state[pos] = (
                new_state[pos], new_state[zero]
            )

            new_state = tuple(new_state)

            if new_state not in path:
                result = dfs(
                    new_state, limit - 1, path + [new_state]
                )

                if result is not None:
                    return result

    return None


def ids(start):
    depth = 0

    while True:
        print("Trying depth:", depth)

        result = dfs(start, depth, [start])

        if result is not None:
            return result

        depth += 1


def display(solution):
    for state in solution:
        for i in range(0, 9, 3):
            print(state[i:i + 3])
        print()


start = tuple(map(int, input("Enter 9 numbers (0 for blank): ").split()))

solution = ids(start)

print("\nSolution found!")
print("Number of moves:", len(solution) - 1)
display(solution)
```

Vacuum Cleaner - Simple Reflex Agent


```python

room = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"


def simple_reflex_agent(location, room):

    if room[location] == "Dirty":
        return "Suck"

    elif location == "A":
        return "Move Right"

    else:
        return "Move Left"


while True:

    print("\nLocation:", location)
    print("Room A:", room["A"])
    print("Room B:", room["B"])

    action = simple_reflex_agent(location, room)

    print("Action:", action)

    if action == "Suck":
        room[location] = "Clean"

    elif action == "Move Right":
        location = "B"

    elif action == "Move Left":
        location = "A"

    if room["A"] == "Clean" and room["B"] == "Clean":
        print("All rooms are clean!")
        break
```

 Vacuum Cleaner - Goal Based Agent

```python


room = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"


def goal_based_agent(location, room):

    # Goal: Both rooms should be clean

    if room[location] == "Dirty":
        return "Suck"

    elif room["A"] == "Dirty":
        return "Move to A"

    elif room["B"] == "Dirty":
        return "Move to B"

    else:
        return "Stop"


while True:

    print("\nLocation:", location)
    print("Room A:", room["A"])
    print("Room B:", room["B"])

    action = goal_based_agent(location, room)

    print("Action:", action)

    if action == "Suck":
        room[location] = "Clean"

    elif action == "Move to A":
        location = "A"

    elif action == "Move to B":
        location = "B"

    elif action == "Stop":
        print("Goal Achieved! All rooms are clean.")
        break
```

```python
import heapq

initial = [
    [2, 8, 3],
    [1, 6, 4],
    [7, 0, 5]
]

goal = [
    [1, 2, 3],
    [8, 0, 4],
    [7, 6, 5]
]



def misplaced(state):
    count = 0

    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != goal[i][j]:
                count += 1

    return count


# Manhattan Distance Heuristic
def manhattan(state):
    distance = 0

    for i in range(3):
        for j in range(3):
            tile = state[i][j]

            if tile != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == tile:
                            distance += abs(i - x) + abs(j - y)

    return distance


# Generate possible next states
def get_neighbors(state):
    neighbors = []

    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                row, col = i, j

    moves = [(1, 0), (-1, 0), (0, 1), (0, -1)]

    for dr, dc in moves:
        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [r[:] for r in state]

            new_state[row][col], new_state[nr][nc] = (
                new_state[nr][nc],
                new_state[row][col]
            )

            neighbors.append(new_state)

    return neighbors


# A* Search
def astar(heuristic):
    start = tuple(map(tuple, initial))
    target = tuple(map(tuple, goal))

    pq = []
    heapq.heappush(pq, (heuristic(initial), 0, start, [initial]))

    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state == target:
            print("Solution found!")
            print("Number of moves:", g)

            for step, board in enumerate(path):
                print("\nStep", step)
                for row in board:
                    print(row)

            return

        if state in visited:
            continue

        visited.add(state)

        for neighbor in get_neighbors([list(row) for row in state]):
            next_state = tuple(map(tuple, neighbor))

            if next_state not in visited:
                new_g = g + 1
                new_f = new_g + heuristic(neighbor)

                heapq.heappush(
                    pq,
                    (new_f, new_g, next_state, path + [neighbor])
                )

    print("No solution found.")


# Choose heuristic
print("1. Misplaced Tiles")
print("2. Manhattan Distance")

choice = int(input("Enter your choice: "))

if choice == 1:
    astar(misplaced)
elif choice == 2:
    astar(manhattan)
else:
    print("Invalid choice")
```

N-Queen Problem Hill Climbing

```python
def print_grid(state):
    n = len(state)

    print("    ", end="")
    for col in range(n):
        print(col, end="   ")
    print()

    for row in range(n):
        print(row, end="   ")
        for col in range(n):
            if state[col] == row:
                print(" Q ", end=" ")
            else:
                print(" . ", end=" ")
        print()
    print()


initial = [3, 1, 2, 0]
final = [1, 3, 0, 2]

print("INITIAL BOARD:")
print_grid(initial)

print("FINAL BOARD:")
print_grid(final)

print("Initial Cost: 2")
print("Final Cost: 0")
```

N-Queen using Stimulation Annealng

```python
import random
import math

def cost(state):
    conflicts = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts


def print_grid(state):
    print("  ", end="")
    for col in range(1, 9):
        print(col, end=" ")
    print()

    for row in range(1, 9):
        print(row, end=" ")
        for col in range(8):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print()


def simulated_annealing():
    current = [random.randint(1, 8) for _ in range(8)]
    initial = current.copy()

    T = 1000
    cooling_rate = 0.99

    print("INITIAL STATE:", initial)
    print("Initial Cost:", cost(initial))
    print("Initial Board:")
    print_grid(initial)

    while T > 0.1:
        if cost(current) == 0:
            break

        next_state = current.copy()
        col = random.randint(0, 7)
        row = random.randint(1, 8)
        next_state[col] = row

        deltaE = cost(current) - cost(next_state)

        if deltaE > 0:
            current = next_state
        elif random.random() < math.exp(deltaE / T):
            current = next_state

        T *= cooling_rate

    print("FINAL STATE:", current)
    print("Final Cost:", cost(current))
    print("Final Board:")
    print_grid(current)

    if cost(current) == 0:
        print("Solution found!")
    else:
        print("No solution found. Run again.")


solution = simulated_annealing()
```
